```python
import pandas as pd

# Cargar los archivos CSV
train_df = pd.read_csv("../data/train.csv")
test_df = pd.read_csv("../data/test.csv")
submission_df = pd.read_csv("../data/submission.csv")

# Ver una vista previa de cada dataframe
train_df.head()
test_df.head()
submission_df.head()
```

```python
genre_names = {
    0: "Acoustic/Folk",
    1: "Alt Music",
    2: "Blues",
    3: "Bollywood",
    4: "Country",
    5: "HipHop",
    6: "Indie Alt",
    7: "Instrumental",
    8: "Metal",
    9: "Pop",
    10: "Rock",
}

print(f"Filas de entrenamiento: {len(train_df):,}")
print(f"Columnas: {train_df.shape[1]}")
print("Nombres de columnas:")
print(train_df.columns.tolist())

print("\nValores faltantes por columna:")
missing = train_df.isna().sum()
display(missing[missing.gt(0)].rename("faltantes").to_frame())

class_summary = (
    train_df.groupby("Class")
    .agg(
        canciones=("Class", "size"),
        popularidad_media=("Popularity", "mean"),
        baile_medio=("danceability", "mean"),
        energia_media=("energy", "mean"),
        acustica_media=("acousticness", "mean"),
        instrumentalidad_media=("instrumentalness", "mean"),
        habla_media=("speechiness", "mean"),
    )
)
class_summary.index = class_summary.index.map(genre_names)
class_summary["porcentaje"] = class_summary["canciones"] / len(train_df) * 100

display(class_summary.round(2).sort_values("canciones", ascending=False))
```
